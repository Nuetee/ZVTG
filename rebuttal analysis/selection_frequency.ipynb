```python
import json
import matplotlib.pyplot as plt    

# JSON 파일로 저장
with open("TAG_activitynet_Segment_frequency.json") as f:
    data = json.load(f)
```

```python
import numpy as np  # y축 눈금 간격 조정을 위해 추가
plt.figure(figsize=(8, 7))
# 그래프 설정
plt.grid(axis='y', linestyle='-', zorder=0)
plt.bar(data["keys"], data["percentages"], width=28, align='center', color='#4472c4', edgecolor='None', zorder=3)

# 그래프 설정
plt.xlabel("Relative Offset(%) to Ground Truth Segment", fontsize=22)
plt.ylabel("Proportion of Selections(%)", fontsize=22)
# plt.title("ActivityNet Captions", fontsize=16)
plt.xticks(data["keys"], rotation=45, fontsize=22)  # x축 눈금 설정
y_min, y_max = 0, 22
y_ticks = np.arange(y_min, y_max, 10)
plt.yticks(y_ticks, fontsize=22)
plt.ylim(0, 22)  # 최소 10% 이상 보이도록 설정
plt.tight_layout()

# 그래프 표시
plt.show()

plt.close()  # 메모리 해제
```
